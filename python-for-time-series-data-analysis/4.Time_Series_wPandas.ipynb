```python
def first_day(entry):
    if len(entry):
        return entry[0]
```

```python
df.resample(rule="YE").apply(first_day)
```

### Rolling and Expanding

```python
import pandas as pd
```

```python
%matplotlib inline
```

```python
df = pd.read_csv("Data/starbucks.csv", parse_dates=True, index_col="Date")
```

```python
df.head(30)
```

```python
df["Close"].plot(figsize=(16, 8))
```

```python
df.rolling(window=7).mean().head(30)
```

```python
df["Close"].plot(figsize=(12, 5))
df.rolling(window=7).mean()["Close"].plot()
```

```python
df["Close"].plot(figsize=(12, 5))
df.rolling(window=30).mean()["Close"].plot()
```

```python
df["Close:30DayMean"] = df["Close"].rolling(window=30).mean()
```

```python
df.head(10)
```

```python
df[["Close", "Close:30DayMean"]].plot(figsize=(16, 8))
```

```python
df["Close"].expanding().mean().plot(figsize=(12, 5))
```

```python
df["CloseExpanding"] = df["Close"].expanding().mean()
```
